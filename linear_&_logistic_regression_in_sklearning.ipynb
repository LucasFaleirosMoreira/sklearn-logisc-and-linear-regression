```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)
ages = np.random.randint(low=15, high=70, size=40)

ages
```

```python
labels = []
for age in ages:
  if age < 30:
    labels.append(0)
  else:
    labels.append(1)

#random swap
for i in range(0, 3):
  r = np.random.randint(0, len(labels) - 1)
  if labels[r] == 0:
    labels[r] = 1
  else:
    labels[r] = 0
```

```python
plt.scatter(ages, labels, color="red")
plt.show()
```

```python
from sklearn.linear_model import LinearRegression

model = LinearRegression()
model.fit(ages.reshape(-1, 1), labels)
```

```python
#y = m.x + b
m = model.coef_[0]
b = model.intercept_

m, b
```

```python
#m = peso de cada neuronio de entrada
#b = bias
#x = entradas
#y = limiar

#y = m.x + b
#y - b = m.x
#(y - b)/m = x

limiar_idade = (0.5 - b) / m

plt.plot(ages, ages * m + b, color = "blue")
plt.plot([limiar_idade, limiar_idade], [0, 0.5], '--', color="green")
plt.scatter(ages, labels, color="red")
plt.show()
```

```python
#utilizando uma função logistica

import math

def sigmoid(x):
  a = []
  for item in x:
    a.append(1/(1+math.exp(-item)))
  return a

x = np.arange(-10, 10., 0.2)
sig = sigmoid(x)

plt.plot(x, sig)
plt.show()
```

Classificador sigmóide

```python
from sklearn.linear_model import LogisticRegression

model = LogisticRegression()
model.fit(ages.reshape(-1, 1), labels)

m = model.coef_[0][0]
b = model.intercept_[0]

x = np.arange(0, 70, 0.1)
sig = sigmoid(m*x + b)

limiar_idade = abs(b / m)

plt.scatter(ages, labels, color="red")
plt.plot([limiar_idade, limiar_idade], [0, 0.5], '--', color='green')
plt.plot(x, sig)
plt.show()
```
